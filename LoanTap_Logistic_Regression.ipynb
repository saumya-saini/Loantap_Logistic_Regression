Quantifying Outliers using IQR

```python
# Calculate the number and percentage of potential outliers

outlier_summary = []

for col in numerical_cols:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_limit) |
        (df[col] > upper_limit)
    ][col].count()

    outlier_percentage = (outliers / len(df)) * 100

    outlier_summary.append({
        'Feature': col,
        'Outlier Count': outliers,
        'Outlier Percentage': outlier_percentage
    })

outlier_df = pd.DataFrame(outlier_summary)

outlier_df.sort_values(
    by='Outlier Percentage',
    ascending=False
)
```

### Outlier Detection Using the IQR Method

The Interquartile Range (IQR) method was used to identify potential outliers in numerical variables. Observations below Q1 − 1.5 × IQR or above Q3 + 1.5 × IQR were flagged.

Key observations:

1. Public records (14.58%) and public record bankruptcies (11.39%) have the highest proportion of flagged observations. This is largely due to their zero-concentrated distributions, where the first and third quartiles are both zero.
2. Revolving balance (5.37%) and annual income (4.22%) also contain a noticeable proportion of potential outliers, consistent with their positively skewed distributions.
3. Installment (2.84%), open accounts (2.60%) and total accounts (2.15%) contain additional upper-end observations.
4. DTI, loan amount and revolving utilization have relatively few observations flagged by the IQR method.

These findings indicate that several numerical variables contain extreme or unusual values. However, IQR-based outliers are not necessarily erroneous observations. Since extreme financial characteristics may carry useful information for credit-risk assessment, they will not be removed automatically. Further investigation will be performed during preprocessing.

Chi Square Test

Null Hypothesis (H₀):

There is no significant association between employment length and loan status.



Alternative Hypothesis (H₁):

There is a significant association between employment length and loan status.

```python
# Contingency Table
import pandas as pd

contingency_table = pd.crosstab(
    df['emp_length'],
    df['loan_status']
)

contingency_table
```

```python
from scipy.stats import chi2_contingency

chi2, p_value, dof, expected = chi2_contingency(contingency_table)

print("Chi-square statistic:", chi2)
print("P-value:", p_value)
print("Degrees of freedom:", dof)
```

```python
import numpy as np

n = contingency_table.values.sum()

cramers_v = np.sqrt(
    chi2 / (n * min(contingency_table.shape[0]-1,
                    contingency_table.shape[1]-1))
)

print("Cramer's V:", cramers_v)
```

```python
# compare imputation methods
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error

# Select rows where mort_acc is not missing
observed_data = df[df['mort_acc'].notna()][
    ['total_acc', 'mort_acc']
]

# Divide observed data into training and validation sets
train_data, val_data = train_test_split(
    observed_data,
    test_size=0.2,
    random_state=42
)

# Method 1: Overall median
overall_median = train_data['mort_acc'].median()

pred_median = pd.Series(
    overall_median,
    index=val_data.index
)

mae_median = mean_absolute_error(
    val_data['mort_acc'],
    pred_median
)

# Method 2: Group-based median
group_median = train_data.groupby('total_acc')['mort_acc'].median()

pred_group_median = val_data['total_acc'].map(group_median)

pred_group_median = pred_group_median.fillna(overall_median)

mae_group_median = mean_absolute_error(
    val_data['mort_acc'],
    pred_group_median
)

# Method 3: Group-based mean
group_mean = train_data.groupby('total_acc')['mort_acc'].mean()

pred_group_mean = val_data['total_acc'].map(group_mean)

pred_group_mean = pred_group_mean.fillna(
    train_data['mort_acc'].mean()
)

mae_group_mean = mean_absolute_error(
    val_data['mort_acc'],
    pred_group_mean
)

# Compare results
results = pd.DataFrame({
    'Imputation Method': [
        'Overall Median',
        'Group-based Median',
        'Group-based Mean'
    ],
    'MAE': [
        mae_median,
        mae_group_median,
        mae_group_mean
    ]
})

results.sort_values('MAE')
```

```python
# identifying numerical and categorical columns.
numeric_cols = x_train.select_dtypes(include=['int64', 'float64']).columns.tolist()

categorical_cols = x_train.select_dtypes(include=['object']).columns.tolist()

print("Numerical columns:", numeric_cols)
print("\nCategorical columns:", categorical_cols)
```

```python
# Rechecking dti feature
x_train['dti'] = x_train['dti'].replace([9999, 1622], np.nan)
x_test['dti'] = x_test['dti'].replace([9999, 1622], np.nan)
```

```python
print("Missing DTI values in training:", x_train['dti'].isnull().sum())
print("Missing DTI values in testing:", x_test['dti'].isnull().sum())
```

```python
# Numerical preprocessing
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])
```

```python
# Categorical preprocessing
from sklearn.preprocessing import OneHotEncoder

categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(handle_unknown='ignore'))
])
```

```python
# Combine numerical and categorical preprocessing
from sklearn.compose import ColumnTransformer

preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numeric_cols),
        ('cat', categorical_transformer, categorical_cols)
    ]
)
```

## Model Building

```python
from sklearn.linear_model import LogisticRegression

logistic_model = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('classifier', LogisticRegression(max_iter=1000))
])
```

### Precision and Recall

Precision and Recall are important classification evaluation metrics, particularly when working with imbalanced datasets.

**1. Precision: 55.14%**

Precision measures the proportion of correctly identified positive cases among all cases predicted as positive.

In our project, it tells us how many borrowers predicted as Charged Off actually defaulted.

Formula:

Precision = TP / (TP + FP)

= 907 / (907 + 738)

= 55.14%

**Interpretation:** Approximately 55.14% of borrowers predicted as defaulters actually defaulted.

**2. Recall: 5.84%**

Recall measures the proportion of actual positive cases correctly identified by the model.

In our project, it tells us how many actual Charged Off loans were successfully identified.

Formula:

Recall = TP / (TP + FN)

= 907 / (907 + 14,628)

= 5.84%

**Interpretation:** The model successfully identifies only 5.84% of actual defaulters and misses approximately 94.16%.

**Key Observation:**

The model has relatively better precision than recall, but its recall is extremely low.

Although the model can identify some defaulters with reasonable precision, it fails to capture most actual defaults.

Since identifying potential defaulters is important for LoanTap's risk management, improving recall will be an important objective during model improvement.

### F1-Score

F1-score is the harmonic mean of Precision and Recall. It provides a combined measure of the model's ability to identify positive cases while balancing false positive and false negative predictions.

**Formula:**

F1-score = 2 × (Precision × Recall) / (Precision + Recall)

**Calculation:**

= 2 × (0.5514 × 0.0584) / (0.5514 + 0.0584)

= 0.1056 (10.56%)

**Interpretation:**

The model achieved an F1-score of 10.56%, indicating a poor balance between precision and recall.

Although the model has a precision of 55.14%, its recall is only 5.84%. It identifies very few actual loan defaulters.

**Key Observation:**

The low F1-score highlights the model's poor ability to identify Charged Off loans. Improving the detection of potential defaulters will be an important objective during model optimisation.
